## 3. Leakage audit

*The same hunt from Week 3, on your final feature set.*

**Methodology for Leakage Audit:**

Data leakage occurs when information from the validation or test set is inadvertently used during model training, leading to an overly optimistic evaluation of model performance. This can happen in several ways, such as:

1.  **Target leakage**: Features that are directly or indirectly derived from the target variable itself, or become available only *after* the target is known.
2.  **Train-test contamination**: Applying data preprocessing steps (e.g., scaling, imputation, feature engineering) across the entire dataset *before* splitting into train and test sets, thus allowing information from the test set to influence the training set's transformations.
3.  **Time-series leakage**: Using future information to predict past or present events.
4.  **Group leakage**: Not splitting data by logical groups (e.g., users, devices) when predictions need to generalize to new groups.

**Audit Steps for Week-5 Final Feature Set:**

1.  **Review Feature Definitions**: For each feature in the final set, explicitly ask: 'Is this feature available at the time of prediction?' and 'Could this feature contain information that would not realistically be available in a new, unseen data point?'
2.  **Examine Feature Creation Logic**: Scrutinize the code used to generate each feature. Ensure that aggregations, transformations, or derivations are performed *only* on the training data when fitting processors (e.g., `StandardScaler`, `Imputer`) and then applied to both train and test sets.
3.  **Check Time-Dependence**: If the model is time-sensitive, confirm that no future information is used. For example, if a feature is `average_spend_last_7_days`, ensure it's calculated up to `t-1` for a prediction at time `t`.
4.  **Verify Split Integrity**: Re-confirm that the data split (especially the 'honest' split from Section 2) correctly isolates information, particularly for grouped data (e.g., client-wise split).
