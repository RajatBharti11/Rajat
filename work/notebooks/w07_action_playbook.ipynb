## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

### Archetype: Underperforming Evergreen Content

**Action:** Optimize content for search engines (SEO) and user engagement.

**Reason Code:** Content is relevant but not reaching its target audience effectively, leading to low traffic and conversions. This indicates a need for improved visibility and audience retention.

### Archetype: Outdated Factual Content

**Action:** Update factual information, statistics, and examples.

**Reason Code:** Content contains outdated information, which can mislead users and damage credibility. Regular updates ensure accuracy and continued relevance.

### Archetype: High-Performing Content with Engagement Drop-off

**Action:** Refresh calls to action (CTAs) and explore new distribution channels.

**Reason Code:** While initially successful, engagement metrics are declining, suggesting content fatigue or a need for renewed promotion. Optimizing CTAs can re-engage users, and new channels can reach a broader audience.

### Archetype: Content with Low Readability/Clarity

**Action:** Simplify language, improve structure, and add visual aids.

**Reason Code:** Users are abandoning content quickly due to complex language or poor organization. Enhancing readability can improve user experience and information retention.

### Archetype: Duplicate or Redundant Content

**Action:** Consolidate or repurpose similar content pieces.

**Reason Code:** Multiple pieces address the same topic, leading to internal competition and diluted SEO efforts. Consolidating can create a more authoritative resource and improve search rankings.

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

### Human Review Checkpoints:

Before implementing any recommended action from this playbook, a human reviewer (e.g., content editor, subject matter expert) must perform the following checks:

1.  **Brand Voice & Tone Alignment:** Ensure proposed changes maintain the established brand voice and tone. Automated suggestions might optimize for metrics but miss subtle brand nuances.
2.  **Accuracy & Factual Verification (Especially for Critical Content):** For content related to health, finance, legal, or other sensitive topics, a subject matter expert *must* manually verify all factual updates and claims. Automation can flag outdated data but cannot guarantee the accuracy of new information.
3.  **Ethical & Compliance Considerations:** Review content for any potential ethical breaches, biases, or compliance issues (e.g., GDPR, accessibility standards) that automated systems might overlook.
4.  **Strategic Fit with Current Campaigns:** Confirm that proposed content actions align with ongoing marketing campaigns or business objectives. A mechanically optimized piece might contradict a larger strategic push.
5.  **Competitor Analysis (Qualitative):** While quantitative data informs many recommendations, a qualitative review of competitor content can provide context that automation lacks. Is a proposed change still competitive and differentiated?
6.  **SEO Keyword Intent:** Ensure that keyword optimizations genuinely match user search intent and provide value, rather than just stuffing keywords for ranking.

### The No-Go List: What Should Never Be Automated:

Certain content actions carry high risk or require inherently human judgment and should **never be fully automated**:

*   **Generating Entire New Content Pieces (without significant human oversight):** While AI can assist, the creation of original, high-quality, and strategically aligned content requires human creativity, empathy, and understanding of complex narratives.
*   **Altering Core Messaging or Brand Identity:** Changes to foundational brand messages, mission statements, or core values must be approved by human leadership.
*   **Handling Crisis Communications or Sensitive PR:** Automated responses in sensitive situations can lead to severe reputational damage. Human judgment is paramount.
*   **Making Legal or Medical Claims/Recommendations:** Content in regulated industries must always be drafted and reviewed by qualified human experts.
*   **Deleting High-Performing or Strategically Important Content:** While consolidation is good, deleting content with strong historical performance or strategic value without explicit human approval is too risky.
*   **Publishing Content without Final Human Review:** Every piece of content, especially client-facing or public-facing, should undergo a final human review for quality, accuracy, and tone before publication.

## 4. Monitoring / retrain triggers

*What would tell you the recommendations went stale?*

### Monitoring Metrics for Playbook Effectiveness:

To ensure the content action playbook remains relevant and effective, continuous monitoring of key metrics is essential. A significant deviation or trend in these metrics would trigger a review or retraining process for the playbook's recommendations.

1.  **Content Performance Decay:**
    *   **Trigger:** Consistent decline (e.g., 15% drop over 3 months) in organic traffic, engagement rates (time on page, bounce rate), or conversion rates for content pieces *after* playbook actions have been applied.
    *   **What it tells us:** The actions recommended might no longer be effective, or external factors have shifted.

2.  **Sentiment Shift:**
    *   **Trigger:** A sustained negative trend in user comments, social media mentions, or direct feedback regarding content quality or relevance.
    *   **What it tells us:** The content may be failing to meet user expectations, suggesting a misalignment in content strategy or tone.

3.  **Search Algorithm Updates:**
    *   **Trigger:** Major announced or observed shifts in search engine ranking factors (e.g., Google core updates) that significantly impact organic visibility across a broad range of content.
    *   **What it tells us:** SEO-related recommendations in the playbook might become outdated and require re-evaluation based on new algorithm priorities.

4.  **Competitor Content Performance:**
    *   **Trigger:** A sustained increase in competitor content ranking for target keywords or significant gains in their audience engagement, while our own remains stagnant or declines.
    *   **What it tells us:** Our content strategy or optimization tactics may be falling behind industry benchmarks, requiring an update to competitive analysis within the playbook.

5.  **User Search Intent Evolution:**
    *   **Trigger:** Analytics reveal a noticeable shift in the long-tail keywords or questions users are employing to find information, differing from the original intent targeted by existing content.
    *   **What it tells us:** The content no longer fully addresses evolving user needs, necessitating new content creation or significant re-optimisation for new search intents.

### Retraining and Refresh Triggers:

When monitoring metrics indicate a problem, the following actions should be triggered:

*   **Quarterly Review:** A mandatory, in-depth review of the playbook's archetypes and recommended actions every quarter.
*   **Ad-hoc Review for Major Shifts:** Initiate an immediate review if there's a significant external event (e.g., new product launch, major industry regulation, platform policy change) that could invalidate existing content recommendations.
*   **Model Retraining (if applicable):** If any AI/ML models are used to generate content insights or recommendations, these models should be retrained with the latest performance data to adapt to new trends and user behaviors.

```python
import datetime

# This cell is for CODE (numbers, a query, a check).
# Here, we can define simple threshold checks for monitoring metrics that would trigger a review.

# Simulate monitoring metrics
current_organic_traffic_decline = 0.18 # 18% decline over 3 months
threshold_organic_traffic_decline = 0.15 # Trigger if decline > 15%

current_sentiment_score = 3.2 # On a scale of 1-5, lower is worse
threshold_sentiment_score = 3.5 # Trigger if score drops below 3.5

last_algorithm_update = datetime.date(2023, 10, 27)
known_major_update = datetime.date(2024, 1, 15) # Date of a recent major algorithm update

print("--- Monitoring and Retrain Triggers ---")

# Check for content performance decay
if current_organic_traffic_decline > threshold_organic_traffic_decline:
    print(f"ALERT: Organic traffic has declined by {current_organic_traffic_decline*100:.0f}% (exceeds {threshold_organic_traffic_decline*100:.0f}% threshold). TRIGGER PLAYBOOK REVIEW.")
else:
    print(f"Organic traffic decline ({current_organic_traffic_decline*100:.0f}%) is within acceptable limits.")

# Check for sentiment shift
if current_sentiment_score < threshold_sentiment_score:
    print(f"ALERT: Average sentiment score ({current_sentiment_score}) is below threshold ({threshold_sentiment_score}). TRIGGER CONTENT QUALITY REVIEW.")
else:
    print(f"Sentiment score ({current_sentiment_score}) is stable.")

# Check for recent major algorithm updates
if known_major_update > last_algorithm_update:
    print(f"ALERT: A major search algorithm update occurred on {known_major_update}. TRIGGER SEO RECOMMENDATION REVIEW.")
else:
    print("No recent major search algorithm updates detected that would invalidate SEO recommendations.")

print("Monitoring checks complete.")
```
